## HYPOTHESIS TESTING

#### Background:

Bombay hospitality Ltd. operates a franchise model for producing exotic Norwegian dinners throughout New England. The operating cost for a franchise in a week (W) is given by the equation W = $1,000 + $5X, where X represents the number of units produced in a week. Recent feedback from restaurant owners suggests that this cost model may no longer be accurate, as their observed weekly operating costs are higher.

Objective: To investigate the restaurant owners' claim about the increase in weekly operating costs using hypothesis testing.

Data Provided:

•	The theoretical weekly operating cost model: W = $1,000 + $5X

•	Sample of 25 restaurants with a mean weekly cost of Rs. 3,050

•	Number of units produced in a week (X) follows a normal distribution with a mean (μ) of 600 units and a standard deviation (σ) of 25 units

#### 1. State the Hypotheses statement:

```python
## Hypothesis testing (Bombay Hospitality)..

## H0 (Null Hypothesis) : The mean weekly operating cost is equal to the theoretical cost given by the model
## W = $1,000 + $5X

## H1 (Alterenative Hypothesis): The mean weekly operating cost is greater than the theoretical cost
```

#### 2. Calculate the Test Statistic:
#### Use the following formula to calculate the test statistic (t):
#### where:
#### •	ˉxˉ = sample mean weekly cost (Rs. 3,050)
#### •	μ = theoretical mean weekly cost according to the cost model (W = $1,000 + $5X for X = 600 units)
#### •	σ = 5*25 units
#### •	n = sample size (25 restaurants)

```python
## Given values:
sample_mean = 3050
X_mean = 600
n = 25

## Theoretical mean cost from model
mu0 = 100 + 5 * X_mean

## Given standard deviation
sigma = 5 * 25

## Standard error
standard_error = sigma / (n ** 0.5)

## Z-test statistic
z = (sample_mean - mu0) / standard_error

print("Test Statistic (z value): ", z)
```

#### 3. Determine the Critical Value:
#### Using the alpha level of 5% (α = 0.05), determine the critical value from the standard normal (Z) distribution table.

```python
alpha = 0.05 

## For right tailed test at 5% level of significance, critical Zvalue from standard normal table is 1.645
z_critical = 1.645

print("Critical Z value: ", z_critical)
```

#### 4. Make a Decision:
#### Compare the test statistic with the critical value to decide whether to reject the null hypothesis.

```python
if z > z_critical:
    print("Reject the Null Hypothesis")
else:
    print("Fail to reject the Null Hypothesis")
```

#### 5. Conclusion:
#### Based on the decision in step 4, conclude whether there is strong evidence to support the restaurant owners' claim that the weekly operating costs are higher than the model suggests.

```python
if z > z_critical:
    print(r"There is strong evidence to supoort the retaurant owner's claim thet the weekly operating cost is higher than the model suggests.")
else:
    print(r"There is not enough evidence to supoort the retaurant owner's claim thet the weekly operating cost is higher than the model suggests.")
```

```python
## Restaurant owner's claim is that the weekly operating costs are higher than those predicted by the theoretical cost model (W = 1000 + 5X).
## Based on the hypothesis testing results, the calculated Z-test statistic was -2.0
## while the critical Z-value at the 5% significance level for a right-tailed test was 1.645. 
## Since the calculated Z-value is less than the critical value, we fail to reject the null hypothesis. 
## This indicates that there is not enough statistical evidence to support the restaurant owner's claim. 
## Therefore based on the sample data provided the existing operating cost model remains valid 
## There is no significant increase in weekly operating costs.
```

## CHI-SQUARE TEST

#### BACKGROUND

Association between Device Type and Customer Satisfaction

Background:
Mizzare Corporation has collected data on customer satisfaction levels for two types of smart home devices: Smart Thermostats and Smart Lights. They want to determine if there's a significant association between the type of device purchased and the customer's satisfaction level.

Data Provided:
The data is summarized in a contingency table showing the counts of customers in each satisfaction level for both types of devices:

```python
## Satisfaction	      Smart Thermostat	  Smart Light     Total
## Very Satisfied	       50	             70	          120
## Satisfied	           80	             100	      180
## Neutral	               60	             90	          150
## Unsatisfied	           30	             50	          80
## Very Unsatisfied	       20	             50	          70
## Total	               240	             360	      600


## Objective:
## To use the Chi-Square test for independence to determine if there's a significant association between the type of smart home device purchased (Smart Thermostats vs. Smart Lights) and the customer satisfaction level.
```

### Assignment Tasks:(CHI-SQUARE TEST)

#### 1. State the Hypotheses:

```python
## Null Hypothesis (H0) : There is no significant association between the type of smart home device purchased and the customer satisfaction level. 
## (Device type and the satisfaction level are independent)

## Alternative Hypothesis (H1): There is a significant association between the type of smart home device purchased and the customer satisfaction level. 
## (Device type and the satisfaction level are dependent)
```

#### 2. Compute the Chi-Square Statistic:

```python
import numpy as np
```

```python
## observed frequency table..
observed = np.array([
    [50, 70],
    [80, 100],
    [60, 90],
    [30, 50],
    [20, 50]
])
```

```python
## row totals..
row_totals = observed.sum(axis = 1)
print(row_totals)

## column totals..
col_totals = observed.sum(axis = 0)
print(col_totals)

## grand total..
grand_total = observed.sum()
print(grand_total)
```

```python
## calculate the expected frequencies..
expected = np.outer(row_totals, col_totals) / grand_total
print(expected)
```

```python
## calculate chi square statistic.. 
chi_square = ((observed - expected) ** 2 / expected).sum()

print("Chi-Square Statistic: ", chi_square)
```

#### 3. Determine the Critical Value:
#### Using the significance level (alpha) of 0.05 and the degrees of freedom (which is the number of categories minus 1)

```python
alpha = 0.05 
print("alpha: ", alpha)

## degrees of freedom = (rows - 1)(columns - 1)
df = (5 - 1) * (2 - 1)
print("Degrees of freedom: ", df)

## critical value from chi-square table for df=4 at 5% significance level is 9.488
chi_critical = 9.488
print("Critical Value: ", chi_critical)
```

#### 4. Make a Decision:
#### Compare the Chi-Square statistic with the critical value to decide whether to reject the null hypothesis.

```python
if chi_square > chi_critical:
    print("Reject the Null Hypothesis")
else:
    print("Failed to reject the Null Hypothesis")
```

```python
## conclusion
if chi_square > chi_critical:
    print("There is a significant association between the type of smart home device purchased and the customer satisfaction level.")
else:
    print("There is no significant association between the type of smart home device purchased and the customer satisfaction level.")
```

```python
## Conclusion:
## Based on the Chi-Square Test for Independence the calculated Chi-Square statistic was 5.638
## while the critical Chi-Square value at a 5% significance level with 4 degrees of freedom was 9.488. 
## Since the calculated Chi-Square value is less than the critical value, we fail to reject the null hypothesis. 
## This indicates that there is insufficient statistical evidence to conclude that a significant association exists between the type of smart home device purchased (Smart Thermostats or Smart Lights) and the customer satisfaction level.
## Therefore customer satisfaction appears to be independent of the type of smart home device purchased based on the given sample data.
```
