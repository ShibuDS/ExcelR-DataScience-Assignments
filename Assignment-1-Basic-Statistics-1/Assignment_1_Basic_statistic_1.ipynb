## Boxplot 

```python
import seaborn as sns
```

```python
for i in num_cols:
    Q1 = df[i].quantile(0.25)
    Q3 = df[i].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    min_val = df[i].min()
    max_val = df[i].max()

    ## detect outliers..
    outliers = df[(df[i] < lower_bound) | (df[i] > upper_bound)][i]

    print(f"\nColumn: {i}")
    print(f"Q1 : {Q1}")
    print(f"Q3 : {Q3}")
    print(f"IQR : {IQR}")
    print(f"Lower Bound : {lower_bound}")
    print(f"Upper Bound : {upper_bound}")
    print(f"Minimum Value : {min_val}")
    print(f"Maximum Value : {max_val}")
    print(f"Potential Outiers : {len(outliers)}")
    if len(outliers) > 0:
        print("Potential Outliers:")
        print(outliers.values)
    
    plt.figure(figsize = (7, 2))
    plt.boxplot(df[i].dropna(), vert = False)
    plt.title(f"Boxplot of {i}")
    plt.show()

    print("*"*100)
```

```python
## Interpretation:

## The boxplot for volume shows that most values are concentrated around low ranges with a few points above the upper bound (whisker). This means most transactions have small quantities and only a few have high volume sales (potential outliers). 

## For Average price there are many outliers on the higher side and the box itself is near the lower range. This shows most of the products are low priced but some products have very high price.

## Total sale value also shows a large number of upper outliers. The spread of the box is small compared to the extreame values, which means most sales are of lower value and only a few transactions contributes very high sales.

## Discount Amount has several high outliers and the box is near the lower range. This means most discounts are small but some transations have very large discount amounts.

## Discount Rate(%) has outliers towards the lower range. The values are closly packed which shows that discount percentage is almost similar for most of the records.

## Net Sales Value shows many extreme values on the upper side similar to the Total sales value. This shows that most transactions generate low revenue but a few generate very high revenue.

## Overall potential outliers are mostly present in all the columns. Volume and Discount Rate(%) are comparitively lot more stable as compared to other numerical columns.  





## IQR: Inter Quartile Range. It measure the spread of middle 50% datapoints of any numerical column of the dataset.

## From the boxplot that we have created, its not easy to find out exact IQR (Inter quartile range) just by seeing it because we wont be able to find out the exact values of Q3 (75th percentile, upper edge of the box, 3rd quartile) and Q1 (25th percentile, lower edge of the box, 1st quartile). Thats why we manually calculated Q1, Q3, IQR, lower and upper bound, minimum and maximum values to get a clear picture about potential outliers. 

## Formula to find out [IQR = Q3 - Q1]. 
## We just have to find out Q3 and Q1 for each numerical columns and we will be easily able to find out IQR for each column. 
## Biggest perks of IQR is that it is not effected by the outliers because it clearly ignores the outliers. 
## QR is the powerful method to detect strict potential outliers. 
## Any datapoint that falls below [Q1 - (1.5*IQR)] or above [Q3 + (1.5*IQR)] is typically considered as potential outlier. 
## If the dataset is skewed, IQR is considered as the most appropriate measure of variability. 

## A lower IQR indicates that the data is more stable and less spreadout, as the middle values are closely grouped together.
## On the other hand a lager IQR suggests higher variability in the data. In this data, except Volume and Discount Rate(%) rest of the other numerical columns have higher variability. 
## Similarly if the difference between the upper bound and the maximum value is very large, the data point can be considered as an extreme outlier. The same thing goes for lower bound and minimum values as well (extreme lower outliers).

## The number of potential outliers ranges from 24 to 60. 
## Avg Price has 60 potential outliers (maximum) while Discount Amount has least number ouf outliers (24).
```

### Standardization of Numerical Variables

```python
## Explain the concept of standardization (z-score normalization) ?

## - Standardization (z-score normalization) is done to bring all the columns into same scale. Here we convert whole dataset in term z scores using the formula : [ z = (x-mu)/sigma ]. Values get transformed in terms of standard deviation.  

## Here x = original value, mu = mean of the column and sigma = standard deviation. After standardization: Mean becomes 0. Standard deviation becomes 1. Roughly 99.7% of the data points naturally falls between -3 and +3, values outside these ranges are probably potential outliers. Shape of the distribution does not change. Helps when columns have different units and ranges. 
```

```python
print(num_cols)
```

```python
df_std = df.copy()
```

```python
for i in num_cols:
    mean = df[i].mean()
    std = df[i].std()
    df_std[i] = (df[i] - mean) / std
```

```python
## All the values got converted in terms of z score, standardization. 
print(df_std.head())
```

```python
## after standardization mean became almost 0
df_std[num_cols].mean()
```

```python
## after standardization standard deviation became 1
df_std[num_cols].std()
```

```python
## Before and after distribution : 
```

```python
df[num_cols].hist(figsize = (12,8))
plt.show()
```

```python
df_std[num_cols].hist(figsize = (12,8))
plt.show()
```

```python
## Interpretation : 

## After z-score standardization, the mean of all the numerical columns became approximately 0 and the standard deviation became 1. This confirms that the data has been successfully standardized. 

## Before standardization the numerical columns were in different units and range. But after standardization all the variable were brought to same scale, that is in terms of z scores (usually between -3 and +3). 

## While comaparing the histograms we can clearly see that the before and after stadardization distribution of the data, histograms are exactly same, no changes at all, before and after. So it proves that the standardization process does change the distribution of any coloum at all.

## Now why standardization is important ? It is important because some of the machine learning algoriths are quite sesitive to the magnitute of the values. They might tend to underestimate columns with lesser range and units.  
```

### Conversion of Categorical Data into Dummy Variables

```python
## Discuss the need for converting categorical data into dummy variables (one-hot encoding)
## - Machine learning algorith works with numerical data and cannot directly understand categorical values. In our dataset columns like City, Brand, BU, Model etc are in categorical form. These columns contains lables not numbers. If we give them directly to an ML model the algorithm might not work properly at all. 

## We are going to apply one hot encoding (Dummy variables). It converts each category into seperate binary column. In this way all the category gets equal say in model building process and become suitable for ML algorithms. 

## Importance of one hot encoding : 
## Converts categorical data into numeric format. Prevents misinterpretation of the category importance. Makes dataset model ready. Hence it improves accuracy and reliability of ML models.
```

```python
cat_cols
```

```python
## Applying one hot encoding 
df_encoded = pd.get_dummies(df, columns = cat_cols, drop_first = False) 
```

```python
print('Shape of df:',df.shape)
print('Shape of df_encoded:',df_encoded.shape)
```

```python
## Interpretation: 

## Categorical variables cannot be directly used in most machine learning algorithms because they work with numerical data. Therefore the categorical columns were converted into dummy variables using one hot encoding. 

## Here each category is transformed into seperate binary column containing values 0 and 1. A value of 1 indicates the presence of that category in a particular row. 0 indicates its absence. 

## After applying one hot encoding the dataset increased in the number of columns, but it became suitable for machine learning models since all the variable are now in numerical format. 

## Internally True is treated as 1. False is Treated as 0. 

## drop_first = Fasle, will keep all the categories. 
```

```python
## 	Displaying a portion of the transformed dataset 
df_encoded.head()
```

```python
## The values appears as True or False instead of 1 or 0. 
## Although most of the machine learning algorithms algorithms iterpret these as 1 or 0 automatically.
## But we can easily convert them in terms of 1 or 0 using: df_encoded = df_encoded.astype(int)
## Converting them explicitly into numeric form  improves consistency, readabilty and compatibility across different models. 
```
