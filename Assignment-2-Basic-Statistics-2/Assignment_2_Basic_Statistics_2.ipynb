# Estimation and Confidence Intervals

### Background
In quality control processes, especially when dealing with high-value items, destructive sampling is a necessary but costly method to ensure product quality. The test to determine whether an item meets the quality standards destroys the item, leading to the requirement of small sample sizes due to cost constraints. 


### Scenario
A manufacturer of print-heads for personal computers is interested in estimating the mean durability of their print-heads in terms of the number of characters printed before failure. To assess this, the manufacturer conducts a study on a small sample of print-heads due to the destructive nature of the testing process.
Data
A total of 15 print-heads were randomly selected and tested until failure. The durability of each print-head (in millions of characters) was recorded as follows:
1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29

```python
## Lets import required libraries..
import numpy as np
from scipy import stats
```

```python
## Durability data (in millions of character)..
durability = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]
```

## Assignment Tasks

### a. Build 99% Confidence Interval Using Sample Standard Deviation
Assuming the sample is representative of the population; construct a 99% confidence interval for the mean number of characters printed before the print-head fails using the sample standard deviation. Explain the steps you take and the rationale behind using the t-distribution for this task.

```python
## Lets calculate sample statistics..
n = len(durability) ## for sample size

sample_mean = np.mean(durability) ## Sample mean

sample_std = np.std(durability, ddof = 1) ## Sample standard deviation (degrees of freedom = 1 for sample standard deviation)

print("Sample Size:", n)
print("Sample Mean:", round(sample_mean, 4))
print("Sample Standard Deviation:", round(sample_std, 4))
```

```python
sample_mean
```

```python
sample_std
```

```python
## Now lets define Confidence Level..
confidence = 0.99 ## Confidence level

alpha = 1 - confidence  ## Alpha value

df = n - 1 ## Degrees of freedom

print("Degrees of Freedom:", df)
```

```python
## Now lets calculate t critical value..
t_critical = stats.t.ppf(1 - alpha/2, df)

print("t-critical Value:", t_critical)
```

```python
## now lets caculate Margin of error..
margin_error = t_critical * (sample_std / np.sqrt(n))
print("Margin of Error:", margin_error)
```

```python
## Now that we have both sample mean and margin of error, lets calculate confidence interval..

lower_limit = sample_mean - margin_error
upper_limit = sample_mean + margin_error

print("99% Confidence Interval:")
print("Lower Limit =", round(lower_limit, 3))
print("Upper Limit =", round(upper_limit, 3))
```

```python
## Conclusion..
print(f"""We are 99% confident that the true mean durability of the print-heads lies between {round(lower_limit, 3)} and \
{round(upper_limit, 3)} million characters""")
```

```python
## Reason behind using t-Distribution..
print("""Reason for using t-distribution:
1. Sample size is small (n < 30)
2. Population standard deviation is unknown""")
```

### b. Build 99% Confidence Interval Using Known Population Standard Deviation
If it were known that the population standard deviation is 0.2 million characters, construct a 99% confidence interval for the mean number of characters printed before failure.

```python
## Given population deviation..
population_std = 0.2
```

```python
## Lets find z-Critical value..
z_critical = stats.norm.ppf(1 - alpha/2)
print("z-critical Value:", round(z_critical, 3))
```

```python
## Now lets calculate margin of error..
margin_error_z = z_critical * (population_std / np.sqrt(n))

print("Margin of Error:", round(margin_error_z, 4))
```

```python
## Now we can calculate Confidence interval using sample mean and margin of error (for z)..
lower_limit_z = sample_mean - margin_error_z
upper_limit_z = sample_mean + margin_error_z

print("99% Confidence Interval using known Population Standard Deviation:")
print("Lower Limit =", round(lower_limit_z, 3))
print("Upper Limit =", round(upper_limit_z, 3))
```

```python
## Conclusion..
print(f"""We are confident that the true mean durability of the print-heads lies between {round(lower_limit_z, 3)} \
and {round(upper_limit_z, 3)} million characters""")
```

```python
## Reason for using z-distribution..
print("""Reason for using z-distribution: \
Population standard deviation is known""")
```
