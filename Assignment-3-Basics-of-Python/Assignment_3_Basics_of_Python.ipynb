### Exercise 1: Prime Numbers

Write a Python program that checks whether a given number is prime or not.
A prime number is a natural number greater than 1 that has no positive divisors other than 1 and itself.

```python
num = int(input("Enter a number: ")) ## first we take user input..

if num > 1: # prime numers are greater than 1 and has only 2 divisor, 1 and the number itself..
    for i in range(2, num):
        if num % i == 0:  ## checking divisibility from 2 to (num-1).. 
            print(num, "is not a prime number")
            break
    else:
        print(num, "is a prime number")
else:  ## if num is less than
    print(num, "is not a prime number")

##  my idea here was to 1st take user input and than check if it has any divisor other than 1 and itself..
## if yes, it is not a prime number.. 
```

### Exercise 2: Product of Random Numbers

```python
import random ## to gerenate random number we use random library
```

```python
## generating 2 random numbers (between 0 and 20)..
num1 = random.randint(1, 20)
num2 = random.randint(1, 20)

## Printing num1 and num2
print("num1: ", num1)
print("num2: ", num2)

## calculating the actual product of newly generated 2 integer numbers..
actual_product = num1 * num2

## creating an user input, asking to enter the multiplication of newly generated 2 integers.. 
user_input = int(input("What is the product of num1 and num2? :"))

if user_input == actual_product:
    print("Answer is absolutly correct!!")
else:
    print("Wrong answer. Correct answer is: ", actual_product)

## This program generates two integers and then asks the user to calculate the products..
## Then it compares the user answer and the actual product.. 
## At last it will display wheather the answer is correct or not.. 
```

## Exercise 4: Word counter

```python
def word_counter(n): ## creating and defining a function name so that it can be reused through out the code multiple times..
    words = n.split() ## this will convert a text into list of words in it..
    result = {} ## creating empty dictionary to store the final output..
    for i in words: 
        if i in result:
            result[i] += 1
        else:
            result[i] = 1
    return result 

## This function splits the text into words, store each word in a dictionary, counts how many times it appears..
## prints the number of time it appears (for every word) in a given input text..
```

```python
txt = "Python is a programming lauguage. Python is case sesitive."
word_counter(txt)
```

## Exercise 5: Check for Palindrome

```python
def check_palindrome(n):
    n = n.lower()
    n = n.replace(" ", "")
    return n == n[::-1]

## I created a function which converts text to lowercase 1st, then removes any spaces if present..
## and compares it with it's own reverse..  
```

```python
check_palindrome("racecar")
```
